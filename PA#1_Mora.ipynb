## A. Word Rotation Problem

```python
def rotate_word(text):
    return text[1:] + text[0]
```

```python
print(rotate_word("Justie"))
```

```python
print(rotate_word("Renzo"))
```

```python
print(rotate_word("Mora"))
```

```python
print(rotate_word("OK"))
```

## B. Username Builder Problem

```python
def make_username(first_name, last_name):
    clean_first = first_name.lower().replace(" ", "")
    clean_last = last_name.lower().replace(" ", "")
    return clean_first + "." + clean_last
```

```python
print(make_username("Ada", "Lovelace"))
```

```python
print(make_username("Justie", "Mora"))
```

```python
print(make_username("Lebron", "James"))
```

## C. Bookend Swap Problem

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last] + middle + [first]
```

```python
print(swap_bookends([1, 2, 3, 4, 5, 6]))
```

```python
print(swap_bookends(["red", "green", "blue"]))
```

```python
print(swap_bookends([7, 6]))
```
